# Task 1: Product Collections (List & Tuples)

```python
## Creating a list named products
```

```python
products = ["Lipstick", "Cornflakes", "Biscuit", "Towel", "Shoes", "Bread"]
```

```python
## Creating a tuple named sample_product
sample_product = ("Bread", 50, "Food")
```

```python
## Print the 2nd and last product from the products list.
print(products[1])
print(products[-1])
```

```python
## Append two new product names to products and then print the updated list.
products.append("chocolate")
products.append("water bottle")
```

```python
## Convert sample_product into a list, change its price and convert it back to a tuple.
sample_product_list = list(sample_product)
print(sample_product_list)
sample_product_list[1] = 70
print(sample_product_list)
sample_product = tuple(sample_product_list)
print(sample_product)
```
