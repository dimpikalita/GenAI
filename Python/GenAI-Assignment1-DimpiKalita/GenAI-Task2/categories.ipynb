#  Task 2: Categories (Sets)

## Create a set of categories

```python
products = ["Lipstick", "Cornflakes", "Biscuit", "Towel", "Shirt", "Bread"]
categories = ["Cosmetics", "Food", "Food", "Cloth", "Cloth", "Food"]

categories_set = set(categories)
print(categories_set)

## Adding a new element to the set
categories_set.add("Electronics")
print(categories_set)

## Adding an existing element does not change the set
categories_set.add("Food")
print(categories_set)
```

```python
## Checking whether a category exists in the set
print("Food" in categories_set)
```

```python
print("Mobiles" in categories_set)
```

```python
## Getting the total number of unique categories
print(len(categories_set))
```
