# Task 3: Product Pricing (Dictionaries)

```python
## Creating dictionary price_dict where keys are product names and values are prices
price_dict = {"Lipstick": 475, "Cornflakes": 143.6, "Biscuit": 20, "Towel": 50, "Shirt": 800, "Bread": 50}
```

```python
## Add a new product with price to price_dict
price_dict["Laptop"] = 50000
print(price_dict)
```

```python
## Update the price of an existing product in price_dict
price_dict["Shirt"] = 750
print(price_dict)
```

```python
## Remove a product by name (handle the case when the product does not exist).
product_to_remove = "Biscuit"
if product_to_remove in price_dict:
    del price_dict[product_to_remove]
    print(f"'{product_to_remove}' has been removed.")
else:
    print(f"'{product_to_remove}' not found in price_dict.")
print(price_dict)
```

```python
## Remove a product by name (handle the case when the product does not exist).
product_to_remove = "Lipstick"
if product_to_remove in price_dict:
    del price_dict[product_to_remove]
    print(f"'{product_to_remove}' has been removed.")
else:
    print(f"'{product_to_remove}' not found in price_dict.")
print(price_dict)
```

```python
## Print the average price of all products (use only dictionary operations and basic arithmetic).
average_price = sum(price_dict.values()) / len(price_dict)
print(f"The average price of all products is: {average_price}")
```

```python
## Print the product with both the maximum and minimum prices.
max_price_product = max(price_dict, key=price_dict.get)
min_price_product = min(price_dict, key=price_dict.get)
print(f"The product with the maximum price is: {max_price_product} ({price_dict[max_price_product]})")
print(f"The product with the minimum price is: {min_price_product} ({price_dict[min_price_product]})")
```
