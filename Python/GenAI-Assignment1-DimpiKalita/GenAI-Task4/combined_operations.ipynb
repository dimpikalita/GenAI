# Task 4: Combined Operations

```python
## create a list of tuples named catalog
products = ["Lipstick", "Cornflakes", "Biscuit", "Towel", "Shirt", "Bread"]
price_dict = {"Lipstick": 475, "Cornflakes": 143.6, "Biscuit": 20, "Towel": 50, "Shirt": 800, "Bread": 50}
categories = ["Costmetics", "Food", "Food", "Cloth", "Cloth", "Food"]
catalog = tuple(zip(products, price_dict.values(), categories))

print(catalog)
```

```python
## creat a new dictionary category_to_products
category_to_products = {}
for product, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []
    category_to_products[category].append(product)

print(category_to_products)
```

```python
## Print all products that belong to the category that has the maximum number of products.
category_count = {category: len(products) for category, products in category_to_products.items()}
max_category = max(category_count, key=category_count.get)
print(category_to_products[max_category])
```
