# Task 1: Discount Rules (if / elif / else)

## Variable order_amount_input store the user input. 
## Variable order_amount converts the order_amount_input to integer value.

```python
order_amount_input = input("Enter the order amount: ")
if order_amount_input.isdigit():
    order_amount = int(order_amount_input)
    print(f"Order amount entered: {order_amount}")
else:
    print("Invalid order amount. Order amount should be a number")
```

## discount stores the discount amount applied on the order_amount
## final_amount stores the final order amount after discount.

```python
discount = 0
final_amount = 0
```

## discount rules
## - order__amount >= 2000 -> 15% discount
## - 1500 <= order_amount < 2000 -> 10% discount
## - 1000 <= order_amount < 1500 -> 7% discount
## - otherwise -> 0% discount

```python
if order_amount >= 2000:
    discount = order_amount * (15/100)
elif order_amount >= 1500 and order_amount < 2000:
    discount = order_amount * (10/100)
elif order_amount >= 1000 and order_amount < 1500:
    discount = order_amount * (7/100)
final_amount = order_amount - discount
print(f"The final amount after discount is: Rs{final_amount}")
```

## tax stores fixed 5% tax value which will be added to final_amount after discount

```python
tax = final_amount * 0.05  # 5% of the subtotal after discount
final_total = final_amount + tax
print(f"The sub total is Rs {final_amount}, tax is Rs {tax} and final total is Rs {final_total}")
```
