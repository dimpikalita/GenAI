# Task 2: Process Multiple Orders (for loop)

## orders stores the list of order amounts

```python
orders = [1200, 2500, 800, 1750, 3000]
discount = 0
final_amount = 0
```

## Using a for loop to display the summary table: order_amount -> discount% -> final_amount.

```python
print("Order amount -> Discount -> Final amount")
total_revenue = 0 # Total revenue is the sum of final amount
orders_with_discount = 0 # orders_with_discount stores the count of orders which received discount
for order_amount in orders:
    discount = 0
    if order_amount >= 2000:
        discount = order_amount * 0.15
    elif order_amount >= 1500 and order_amount < 2000:
        discount = order_amount * 0.10
    elif order_amount >= 1000 and order_amount < 1500:
        discount = order_amount * 0.07
    if discount > 0:
        orders_with_discount += 1
    final_amount = order_amount - discount
    total_revenue += final_amount
    print(f"{order_amount} -> {discount} -> {final_amount}")

print(f"\nTotal revenue after discounts: {total_revenue}")
print(f"The number of orders that received a discount is: {orders_with_discount}")
```
