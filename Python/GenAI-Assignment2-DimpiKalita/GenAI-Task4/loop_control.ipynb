## Task 4: Loop control with conditions (break & continue)

```python
daily = [200, 150, 0, 400, 50, -1, 300] # list of daily sales
```

```python
total_sales = 0
for sales in daily:
    if sales == -1:
        print("Corrupted data")
        break
    elif sales == 0:
        print("No sales for the day")
        continue
    else:
        total_sales += sales
print(f"Total sales is : {total_sales}") 

# Total sales will be 800 as the loop stopped after -1 so the last value 300 did not get added to the total sales.
```
